# Task
Compare the DataFrames `df` (loaded from `MELBOURNE_HOUSE_PRICES_LESS.csv`) and `df2` (loaded from `Melbourne_housing_FULL.csv`) by examining their shapes, identifying common and unique columns, comparing data types of common columns, generating descriptive statistics for common numerical columns, and analyzing value counts for common categorical columns.

## Compare DataFrame shapes

### Subtask:

## Identify unique and common columns

### Subtask:
Determine which columns are shared between `df` and `df2`, and which are exclusive to each.


**Reasoning**:
To identify unique and common columns, I will extract column names from both dataframes, then use set operations to find common columns and columns exclusive to each dataframe, and finally print the results.



```python
columns_df = set(df.columns)
columns_df2 = set(df2.columns)

common_columns = list(columns_df.intersection(columns_df2))
unique_to_df = list(columns_df.difference(columns_df2))
unique_to_df2 = list(columns_df2.difference(columns_df))

print(f"Common columns: {common_columns}")
print(f"Columns unique to df: {unique_to_df}")
print(f"Columns unique to df2: {unique_to_df2}")
```

## Compare data types of common columns

### Subtask:
Check if common columns have the same data types across both DataFrames.


**Reasoning**:
To compare the data types of common columns, I will iterate through the `common_columns` list, extract the data types from both DataFrames, print them, and identify any discrepancies.



```python
mismatched_dtypes = []

for col in common_columns:
    dtype_df = df[col].dtype
    dtype_df2 = df2[col].dtype
    print(f"Column: {col}, df dtype: {dtype_df}, df2 dtype: {dtype_df2}")
    if dtype_df != dtype_df2:
        mismatched_dtypes.append(col)

if mismatched_dtypes:
    print(f"\nCommon columns with mismatched data types: {mismatched_dtypes}")
else:
    print("\nAll common columns have matching data types.")
```

## Generate descriptive statistics for numerical columns

### Subtask:
Compare summary statistics (mean, min, max, etc.) for numerical columns present in both DataFrames.


**Reasoning**:
First, I need to identify the numerical columns that are present in both DataFrames from the `common_columns` list. This is necessary before generating descriptive statistics.



```python
numerical_cols_df = df[common_columns].select_dtypes(include=['number']).columns.tolist()
numerical_cols_df2 = df2[common_columns].select_dtypes(include=['number']).columns.tolist()

# Find the intersection of numerical columns to ensure we compare only truly numerical common columns
common_numerical_columns = list(set(numerical_cols_df) & set(numerical_cols_df2))

print(f"Common numerical columns to analyze: {common_numerical_columns}")
```

**Reasoning**:
Now that the common numerical columns have been identified, I will generate and print descriptive statistics for these columns from both `df` and `df2` using the `.describe()` method for comparison.



```python
print("Descriptive statistics for df:")
print(df[common_numerical_columns].describe())
print("\nDescriptive statistics for df2:")
print(df2[common_numerical_columns].describe())
```

**Reasoning**:
Now that the descriptive statistics for common numerical columns have been generated, the next logical step is to analyze the value counts for common categorical columns to continue comparing the DataFrames. To do this, I need to first identify the common categorical columns.



```python
categorical_cols_df = df[common_columns].select_dtypes(include=['object', 'category']).columns.tolist()
categorical_cols_df2 = df2[common_columns].select_dtypes(include=['object', 'category']).columns.tolist()

# Find the intersection of categorical columns to ensure we compare only truly categorical common columns
common_categorical_columns = list(set(categorical_cols_df) & set(categorical_cols_df2))

print(f"Common categorical columns to analyze: {common_categorical_columns}")
```

```python
# Re-examining properties where Landsize is 0.0 and BuildingArea is greater than 0
units_with_zero_landsize = df3[(df3['Landsize'] == 0.0) & (df3['BuildingArea'] > 0.0)]

print("Properties with Landsize = 0.0 and BuildingArea > 0.0 (likely units/apartments):")
print(units_with_zero_landsize[['Type', 'Landsize', 'BuildingArea', 'Price']].head())
print(f"\nNumber of such properties: {len(units_with_zero_landsize)}")
print(f"Property types for these: \n{units_with_zero_landsize['Type'].value_counts()}\n")

print("--- Strategy for Landsize and BuildingArea in Price Prediction ---")
print(
    "1.  **Both Landsize and BuildingArea are crucial for predicting house prices.** "
    "Landsize is typically important for detached houses, while BuildingArea is important for all property types (houses, units, townhouses) as it represents the usable living space."
)

print("2.  **Handling 'Landsize = 0.0' and 'BuildingArea > Landsize' scenarios:**")
print(
    "*   For properties with `Type` 'u' (unit) or `Landsize` == 0.0 (often aligned):\n"
    "    - It's normal for `Landsize` to be 0.0 as the land is shared or not individually titled.\n"
    "    - `BuildingArea` is the more relevant size metric for these properties.\n"
    "    - **Recommendation:** You can create a categorical feature (e.g., 'Is_Unit') to differentiate these, or rely on the 'Type' column. When training models:\n"
    "        a.  **Keep as is**: Tree-based models (like Random Forest, XGBoost) can often handle `Landsize = 0` effectively, learning that it indicates a different property type.\n"
    "        b.  **Feature Engineering**: Create an `is_unit` feature (`df3['is_unit'] = (df3['Type'] == 'u') | (df3['Landsize'] == 0.0)`). Then, consider if `Landsize` should be treated differently (e.g., imputed with a small non-zero value or replaced by `BuildingArea` for units if `Landsize` is zero and `BuildingArea` is non-zero).\n"
    "        c.  **Interaction Terms**: Create interaction features like `Landsize * (1 - is_unit)` and `BuildingArea * is_unit` to emphasize different features based on property type.\n"
    "        d.  **Separate Models**: Build separate price prediction models for houses (Type 'h') and units (Type 'u') as their pricing drivers can be quite different.\n"
)
print(
    "*   For properties where `Landsize` > 0.0 (typically houses):\n"
    "    - Both `Landsize` and `BuildingArea` are critical. `Landsize` accounts for outdoor space, potential for expansion, and location value, while `BuildingArea` is the indoor living space."
)

print(
    "\nConsidering the presence of 'u' (unit) types and `Landsize = 0.0` often aligning with units, the most robust approach is often to leverage the 'Type' column or create an 'Is_Unit' binary feature."
)
```
